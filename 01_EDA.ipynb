```python
print("DESTINATION-LEVEL COST ANALYSIS")
print("=" * 60)

destination_cost_analysis = (
    hotels.groupby("destination_id")["price_per_night"]
    .mean()
    .to_frame("avg_hotel_price")
    .join(
        attractions.groupby("destination_id")["estimated_spend"]
        .mean()
        .to_frame("avg_attraction_spend")
    )
    .join(
        restaurants.groupby("destination_id")["rating"]
        .mean()
        .to_frame("avg_restaurant_rating")
    )
    .join(
        destinations.set_index("destination_id")["avg_daily_budget"]
        .to_frame("avg_daily_budget")
    )
)

print(destination_cost_analysis.round(2))
```

```python
print("OUTLIER CHECK")
print("=" * 60)

outlier_columns = {
    "destinations": ["avg_daily_budget"],
    "attractions": ["entry_fee", "avg_visit_hours", "estimated_spend"],
    "hotels": ["price_per_night", "distance_from_center_km"],
    "restaurants": ["distance_from_center_km"],
    "transportation": ["distance_km", "estimated_time_hours", "estimated_cost"],
    "weather_season": ["rainfall_mm", "humidity"],
    "user_preferences": ["budget", "duration_days", "travelers"],
    "trip_history": ["trip_cost"]
}

for dataset, columns in outlier_columns.items():
    print(f"\n{dataset.upper()}")

    for column in columns:
        df = sheets[dataset]

        Q1 = df[column].quantile(0.25)
        Q3 = df[column].quantile(0.75)
        IQR = Q3 - Q1

        lower = Q1 - 1.5 * IQR
        upper = Q3 + 1.5 * IQR

        outliers = df[
            (df[column] < lower) |
            (df[column] > upper)
        ]

        print(
            f"{column}: {len(outliers)} outliers "
            f"(range: {lower:.2f} to {upper:.2f})"
        )
```

```python
print("OUTLIER ROW INSPECTION")
print("=" * 70)

# 1. Hotel price outlier
Q1 = hotels["price_per_night"].quantile(0.25)
Q3 = hotels["price_per_night"].quantile(0.75)
IQR = Q3 - Q1
lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

print("\nHOTEL PRICE OUTLIERS")
print(
    hotels[
        (hotels["price_per_night"] < lower) |
        (hotels["price_per_night"] > upper)
    ][[
        "hotel_id",
        "destination_id",
        "hotel_name",
        "hotel_type",
        "rating",
        "price_per_night"
    ]]
)


# 2. Restaurant distance outlier
Q1 = restaurants["distance_from_center_km"].quantile(0.25)
Q3 = restaurants["distance_from_center_km"].quantile(0.75)
IQR = Q3 - Q1
lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

print("\nRESTAURANT DISTANCE OUTLIERS")
print(
    restaurants[
        (restaurants["distance_from_center_km"] < lower) |
        (restaurants["distance_from_center_km"] > upper)
    ][[
        "restaurant_id",
        "destination_id",
        "restaurant_name",
        "rating",
        "distance_from_center_km"
    ]]
)


# 3. Transportation outliers
for column in ["distance_km", "estimated_time_hours", "estimated_cost"]:
    Q1 = transportation[column].quantile(0.25)
    Q3 = transportation[column].quantile(0.75)
    IQR = Q3 - Q1
    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    print(f"\nTRANSPORTATION - {column.upper()} OUTLIERS")

    print(
        transportation[
            (transportation[column] < lower) |
            (transportation[column] > upper)
        ][[
            "transport_id",
            "source",
            "destination",
            "transport_type",
            column
        ]]
    )


# 4. Weather humidity outlier
Q1 = weather["humidity"].quantile(0.25)
Q3 = weather["humidity"].quantile(0.75)
IQR = Q3 - Q1
lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

print("\nWEATHER HUMIDITY OUTLIERS")

print(
    weather[
        (weather["humidity"] < lower) |
        (weather["humidity"] > upper)
    ][[
        "destination_id",
        "month",
        "humidity",
        "avg_temperature_c",
        "rainfall_mm",
        "weather_score"
    ]]
)
```

```python
# Create destination feature table

destination_features = destinations[
    [
        "destination_id",
        "destination_name",
        "state_region",
        "country",
        "destination_type",
        "popularity_score",
        "avg_daily_budget"
    ]
].copy()

print("Destination feature table created successfully! 🚀")
print("\nShape:", destination_features.shape)

print("\nPreview:")
display(destination_features.head())
```

```python
# Calculate budget fit score for each destination

destination_features["budget_fit"] = (
    1 - abs(destination_features["avg_daily_budget"] - destination_features["avg_daily_budget"].mean())
    / destination_features["avg_daily_budget"].max()
)

# Keep score between 0 and 1
destination_features["budget_fit"] = (
    destination_features["budget_fit"]
    .clip(0, 1)
    .round(3)
)

print("Budget fit scores created successfully! 💰🚀")

display(
    destination_features[
        [
            "destination_name",
            "avg_daily_budget",
            "budget_fit"
        ]
    ].sort_values("budget_fit", ascending=False)
)
```

```python
# Destination-level budget structure

budget_analysis = (
    destinations[
        ["destination_id", "destination_name", "avg_daily_budget"]
    ]
    .merge(
        hotels.groupby("destination_id")["price_per_night"]
        .mean()
        .rename("avg_hotel_price"),
        on="destination_id"
    )
    .merge(
        attractions.groupby("destination_id")["estimated_spend"]
        .mean()
        .rename("avg_attraction_spend"),
        on="destination_id"
    )
)

budget_analysis["estimated_daily_cost"] = (
    budget_analysis["avg_hotel_price"]
    + budget_analysis["avg_attraction_spend"]
)

print(budget_analysis.round(2))
```

```python
# Transportation cost analysis

transport_analysis = (
    transportation
    .groupby(["source", "destination", "transport_type"])
    [["distance_km", "estimated_time_hours", "estimated_cost"]]
    .mean()
    .reset_index()
)

print("Transportation records:", len(transport_analysis))

print("\nSample transportation data:")
print(transport_analysis.head(15).to_string(index=False))

print("\nAverage cost by transport type:")
print(
    transportation
    .groupby("transport_type")["estimated_cost"]
    .mean()
    .round(2)
)
```

```python
# Restaurant cost analysis

restaurant_analysis = (
    restaurants
    .groupby(["destination_id", "price_level", "food_type"])
    [["rating", "distance_from_center_km"]]
    .mean()
    .reset_index()
)

print("Restaurant pricing groups:", len(restaurant_analysis))

print("\nSample restaurant data:")
print(
    restaurant_analysis
    .head(20)
    .to_string(index=False)
)

print("\nAverage rating by price level:")
print(
    restaurants
    .groupby("price_level")["rating"]
    .mean()
    .round(2)
)

print("\nRestaurant count by price level:")
print(
    restaurants["price_level"]
    .value_counts()
)
```
